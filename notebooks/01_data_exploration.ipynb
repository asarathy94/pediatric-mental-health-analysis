```python
def search_variables(keyword):
    """
    Return dataset column names containing a keyword.

    Parameters
    ----------
    keyword : str
        Word or partial word to search for in column names.
    """
    matches = [
        column
        for column in df.columns
        if keyword.upper() in column.upper()
    ]

    print(f"Found {len(matches)} variables containing '{keyword}':\n")

    for column in matches:
        print(column)

    return matches
```

## Survey Code Audit

Before recoding response values, the unique values and frequencies for each selected variable were reviewed.

This step identifies valid response categories, age-ineligible responses, skipped questions, refusals, unknown responses, and other survey-specific codes. No original values were replaced until their meanings were confirmed.

```python
# Audit each variable for missing values, unique values,
# and common NSCH special response codes

special_codes = [90, 95, 96, 97, 98, 99]

audit_rows = []

for column in analysis_df.columns:
    value_counts = analysis_df[column].value_counts(
        dropna=False
    )

    audit_rows.append({
        "variable": column,
        "data_type": str(analysis_df[column].dtype),
        "rows": len(analysis_df),
        "missing_nan": analysis_df[column].isna().sum(),
        "unique_values": analysis_df[column].nunique(dropna=True),
        "code_90": (analysis_df[column] == 90).sum(),
        "code_95": (analysis_df[column] == 95).sum(),
        "code_96": (analysis_df[column] == 96).sum(),
        "code_97": (analysis_df[column] == 97).sum(),
        "code_98": (analysis_df[column] == 98).sum(),
        "code_99": (analysis_df[column] == 99).sum(),
    })

code_audit_df = pd.DataFrame(audit_rows)

display(code_audit_df)
```

```python
labels_df = pd.read_csv("../docs/NSCH_2024_Survey_Labels_ID.csv")

print(labels_df.shape)
display(labels_df.head())
```

```python
# Rebuild and label the special-code audit in one cell

special_code_columns = [
    "code_90",
    "code_95",
    "code_96",
    "code_97",
    "code_98",
    "code_99",
]

special_code_audit = code_audit_df.loc[
    code_audit_df[special_code_columns].sum(axis=1) > 0
].copy()

labels_lookup = labels_df.rename(
    columns={
        "Name": "variable",
        "Label": "variable_label",
    }
)

special_code_audit_labeled = special_code_audit.merge(
    labels_lookup,
    on="variable",
    how="left",
)

display(
    special_code_audit_labeled[
        [
            "variable",
            "variable_label",
            "code_90",
            "code_95",
            "code_96",
            "code_97",
            "code_98",
            "code_99",
        ]
    ]
)
```

```python
csv_special_codes = [90, 95, 96, 97, 98, 99]

mapping_rows = []

common_columns = [
    col
    for col in analysis_df.columns
    if col in stata_compare_df.columns
    and col != "HHID"
]

for column in common_columns:
    comparison = analysis_df[
        ["HHID", column]
    ].merge(
        stata_compare_df[
            ["HHID", column]
        ],
        on="HHID",
        how="inner",
        suffixes=("_csv", "_stata"),
    )

    comparison = comparison[
        comparison[f"{column}_csv"].isin(csv_special_codes)
    ]

    for raw_code, stata_value in comparison[
        [f"{column}_csv", f"{column}_stata"]
    ].itertuples(index=False, name=None):

        mapping_rows.append({
            "variable": column,
            "csv_code": raw_code,
            "stata_value": str(stata_value),
        })

special_code_mapping = pd.DataFrame(mapping_rows)

if special_code_mapping.empty:
    print("No overlapping special-code mappings were found.")
else:
    special_code_mapping = (
        special_code_mapping
        .groupby(
            ["variable", "csv_code", "stata_value"],
            dropna=False,
        )
        .size()
        .reset_index(name="rows")
        .sort_values(["variable", "csv_code", "stata_value"])
    )

    display(special_code_mapping)
```
