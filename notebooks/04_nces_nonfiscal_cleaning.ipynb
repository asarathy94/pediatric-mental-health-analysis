### Directory Cleaning Summary

- Loaded the NCES CCD 2024–25 District Directory file.

# 3. District Staff

This section inspects and cleans the NCES CCD 2024–25 District Staff data.

The Staff file contains district-level staffing measures by staff category. The cleaned Staff dataset will preserve all analytically useful staffing detail so it can be filtered and related to the Directory and Membership datasets later using `LEAID`.

```python
staff_zip = Path(
    "../data/raw/NCES_CC_Non_Fiscal/Staff/CES_CCD_2024_25_District_Staff.zip"
)

with zipfile.ZipFile(staff_zip, "r") as z:
    print(z.namelist())
```

```python
with zipfile.ZipFile(staff_zip, "r") as z:
    staff_csv = [name for name in z.namelist() if name.endswith(".csv")][0]

    print("Loading:", staff_csv)

    with z.open(staff_csv) as f:
        staff_df = pd.read_csv(f, low_memory=False)

print("Shape:", staff_df.shape)
staff_df.head()
```

### Full Staff Cleaning

The cleaned Staff dataset preserves the complete NCES long-format staffing structure.

Each LEA may appear across multiple rows representing different staff categories, including teachers, school counselors, school psychologists, student support services staff, administrators, paraprofessionals, and other staffing groups.

All staff categories are retained so staffing measures can be filtered and compared later in Tableau, Power BI, or Python.

`DMS_FLAG` is retained to distinguish reported, derived, suppressed, missing, and not-reported values. Unavailable staffing counts remain missing rather than being converted to zero.

`LEAID` is standardized as a string identifier for consistent relationships with the Directory, Membership, and Fiscal datasets.

```python
staff_output = Path(
    "../data/cleaned/NCES_CCD_2024_25_District_Staff_clean.csv"
)

staff_clean.to_csv(staff_output, index=False)

print("Saved:", staff_output)
print("Shape:", staff_clean.shape)
```
