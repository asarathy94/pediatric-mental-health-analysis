### Membership Total Enrollment Cleaning

NCES provides one `Education Unit Total` row for each LEA in the Membership file.

These rows contain the district-wide enrollment total and avoid the need to aggregate across millions of grade, race/ethnicity, and sex subgroup records.

Of the 18,548 LEAs represented in Membership:

- 18,499 have a reported district enrollment total.
- 31 are marked as not reported.
- 12 are suppressed.
- 6 are missing.

Non-reported, suppressed, and missing enrollment values are retained as missing rather than estimated.

```python
membership_clean = membership_totals[
    [
        "SCHOOL_YEAR",
        "LEAID",
        "ST_LEAID",
        "LEA_NAME",
        "ST",
        "STATENAME",
        "FIPST",
        "STUDENT_COUNT",
        "DMS_FLAG"
    ]
].copy()

membership_clean = membership_clean.rename(
    columns={
        "STUDENT_COUNT": "TOTAL_ENROLLMENT",
        "DMS_FLAG": "ENROLLMENT_STATUS"
    }
)

print("Clean membership shape:", membership_clean.shape)
membership_clean.head()
```
