```python
def search_variables(keyword):
    """
    Return dataset column names containing a keyword.

    Parameters
    ----------
    keyword : str
        Word or partial word to search for in column names.
    """
    matches = [
        column
        for column in df.columns
        if keyword.upper() in column.upper()
    ]

    print(f"Found {len(matches)} variables containing '{keyword}':\n")

    for column in matches:
        print(column)

    return matches
```

# Survey Code Audit

Before recoding response values, the unique values and frequencies for each selected variable were reviewed.

This step identifies valid response categories, age-ineligible responses, skipped questions, refusals, unknown responses, and other survey-specific codes. No original values were replaced until their meanings were confirmed.

```python
# Audit each variable for missing values, unique values,
# and common NSCH special response codes

special_codes = [90, 95, 96, 97, 98, 99]

audit_rows = []

for column in analysis_df.columns:
    value_counts = analysis_df[column].value_counts(
        dropna=False
    )

    audit_rows.append({
        "variable": column,
        "data_type": str(analysis_df[column].dtype),
        "rows": len(analysis_df),
        "missing_nan": analysis_df[column].isna().sum(),
        "unique_values": analysis_df[column].nunique(dropna=True),
        "code_90": (analysis_df[column] == 90).sum(),
        "code_95": (analysis_df[column] == 95).sum(),
        "code_96": (analysis_df[column] == 96).sum(),
        "code_97": (analysis_df[column] == 97).sum(),
        "code_98": (analysis_df[column] == 98).sum(),
        "code_99": (analysis_df[column] == 99).sum(),
    })

code_audit_df = pd.DataFrame(audit_rows)

display(code_audit_df)
```

```python
import re
import pandas as pd

# Read the Stata setup file
with open(do_file_path, "r", encoding="utf-8", errors="ignore") as file:
    do_lines = file.readlines()

stata_description_rows = []

# Match lines such as:
# label var k9q96 "Other Adult Child Can Rely On For Advice"

label_var_pattern = re.compile(
    r'^\s*label\s+var(?:iable)?\s+([A-Za-z0-9_]+)\s+"(.*)"\s*$',
    flags=re.IGNORECASE,
)

for line in do_lines:
    match = label_var_pattern.match(line.strip())

    if match:
        stata_description_rows.append({
            "stata_variable": match.group(1).upper(),
            "stata_description": match.group(2).strip(),
        })

stata_description_inventory = (
    pd.DataFrame(stata_description_rows)
    .drop_duplicates(subset="stata_variable")
    .reset_index(drop=True)
)

print(
    "Stata variables with descriptive labels:",
    len(stata_description_inventory)
)

display(stata_description_inventory.head(20))
```

```python
import re
from difflib import SequenceMatcher
import pandas as pd


def normalize_label(text):
    """
    Normalize variable descriptions so small differences in punctuation,
    capitalization, and spacing do not prevent a match.
    """
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = re.sub(r"[_\-/(),:;]", " ", text)
    text = re.sub(r"[^a-z0-9\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


# Keep only the 79 CSV variables without direct Stata name matches
unmatched_csv_variables = variable_match_inventory[
    ~variable_match_inventory["direct_stata_match"]
].copy()

# Create normalized descriptions
unmatched_csv_variables["normalized_csv_label"] = (
    unmatched_csv_variables["variable_label"]
    .apply(normalize_label)
)

stata_description_inventory["normalized_stata_description"] = (
    stata_description_inventory["stata_description"]
    .apply(normalize_label)
)


candidate_rows = []

# Find the three closest Stata descriptions for each unmatched CSV variable
for _, csv_row in unmatched_csv_variables.iterrows():

    csv_label = csv_row["normalized_csv_label"]

    scored_matches = []

    for _, stata_row in stata_description_inventory.iterrows():

        stata_label = stata_row["normalized_stata_description"]

        similarity_score = SequenceMatcher(
            None,
            csv_label,
            stata_label
        ).ratio()

        scored_matches.append({
            "stata_variable": stata_row["stata_variable"],
            "stata_description": stata_row["stata_description"],
            "similarity_score": similarity_score,
        })

    top_matches = sorted(
        scored_matches,
        key=lambda x: x["similarity_score"],
        reverse=True
    )[:3]

    for rank, match in enumerate(top_matches, start=1):
        candidate_rows.append({
            "csv_variable": csv_row["csv_variable"],
            "csv_label": csv_row["variable_label"],
            "candidate_rank": rank,
            "stata_variable": match["stata_variable"],
            "stata_description": match["stata_description"],
            "similarity_score": round(
                match["similarity_score"],
                3
            ),
        })


label_match_candidates = pd.DataFrame(candidate_rows)

display(label_match_candidates.head(30))
```

```python
# Keep only the top-ranked candidate for each unmatched CSV variable

best_label_matches = (
    label_match_candidates[
        label_match_candidates["candidate_rank"] == 1
    ]
    .copy()
)

best_label_matches["match_strength"] = pd.cut(
    best_label_matches["similarity_score"],
    bins=[0, 0.50, 0.70, 0.85, 1.00],
    labels=[
        "Weak",
        "Possible",
        "Strong",
        "Very strong",
    ],
    include_lowest=True,
)

print("Match strength counts:")
print(
    best_label_matches["match_strength"]
    .value_counts(dropna=False)
)

display(
    best_label_matches.sort_values(
        "similarity_score",
        ascending=False
    ).head(30)
)
```

# Primary Outcome: Unmet Mental Health Care

The primary statistical outcome compares children who received needed mental health treatment or counseling with children who needed care but did not receive it.

Children who did not need mental health care are excluded from this outcome-specific analysis.

```python
# Create the primary binary outcome:
# 0 = received needed mental health care
# 1 = needed mental health care but did not receive it

access_df = final_df[
    final_df["MentHCare_24"].isin([1, 2])
].copy()

access_df["unmet_mental_health_care"] = (
    access_df["MentHCare_24"]
    .map({
        1: 0,
        2: 1
    })
    .astype("Int64")
)

print("Children included in mental health access analysis:")
print(len(access_df))

print("\nOutcome counts:")
print(
    access_df["unmet_mental_health_care"]
    .value_counts()
    .sort_index()
)

print("\nOutcome percentages:")
print(
    access_df["unmet_mental_health_care"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
from scipy.stats import chi2_contingency
import numpy as np
import pandas as pd

# Variables to exclude from predictor testing
exclude_vars = {
    "MentHCare_24",
    "MentHDiff_24",
    "mhneeds_24",
    "mhealth_24",
    "unmet_mental_health_care",
}

# Candidate categorical predictors
predictor_vars = [
    column
    for column in access_df.columns
    if column not in exclude_vars
]

results = []

for predictor in predictor_vars:
    analysis_data = access_df[
        [predictor, "unmet_mental_health_care"]
    ].dropna()

    # Skip variables with no meaningful variation
    if analysis_data[predictor].nunique() < 2:
        continue

    contingency_table = pd.crosstab(
        analysis_data[predictor],
        analysis_data["unmet_mental_health_care"]
    )

    # Need at least two outcome columns and two predictor categories
    if contingency_table.shape[0] < 2 or contingency_table.shape[1] < 2:
        continue

    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )

    n = contingency_table.to_numpy().sum()
    min_dimension = min(contingency_table.shape) - 1

    cramers_v = (
        np.sqrt(chi2 / (n * min_dimension))
        if min_dimension > 0
        else np.nan
    )

    results.append({
        "Predictor": predictor,
        "Valid N": n,
        "Categories": contingency_table.shape[0],
        "Chi-square": round(chi2, 3),
        "Degrees of freedom": dof,
        "P-value": p_value,
        "Cramers V": round(cramers_v, 4),
    })

chi_square_results = pd.DataFrame(results)

chi_square_results = chi_square_results.sort_values(
    by=["Cramers V", "P-value"],
    ascending=[False, True]
).reset_index(drop=True)

display(chi_square_results.head(30))
```
