# ACS 2024 Data Cleaning

This notebook prepares selected 2024 American Community Survey (ACS) 5-Year Data Profile tables for analysis.

The ACS tables included are:

- **DP02** — Selected Social Characteristics in the United States
- **DP03** — Selected Economic Characteristics
- **DP04** — Selected Housing Characteristics
- **DP05** — Demographic and Housing Estimates

## Purpose

The goal of this notebook is to:

- load and inspect each ACS table
- review the accompanying column metadata and table notes
- identify variables relevant to the pediatric mental health research questions
- preserve geographic identifiers needed for later analysis and merging
- clean and standardize selected variables
- create processed ACS datasets for later integration with the other project datasets

The original raw ACS files will remain unchanged.

## Combine ACS Curated Tables

This section combines the curated DP02, DP03, DP04, and DP05 tables into one ACS dataset.

The tables are merged using:

- `GEO_ID`
- `NAME`

This preserves the shared geography structure across states and unified school districts while bringing the selected social, economic, housing, and demographic indicators into one analysis-ready file.

```python
acs_curated = (
    dp02_curated
    .merge(
        dp03_curated,
        on=["GEO_ID", "NAME"],
        how="inner"
    )
    .merge(
        dp04_curated,
        on=["GEO_ID", "NAME"],
        how="inner"
    )
    .merge(
        dp05_curated,
        on=["GEO_ID", "NAME"],
        how="inner"
    )
)
```

```python
print("Combined ACS shape:", acs_curated.shape)
print("Duplicate rows:", acs_curated.duplicated().sum())
print("Duplicate GEO_IDs:", acs_curated["GEO_ID"].duplicated().sum())

acs_curated.head()
```

## Standardize State Geography Key

The NSCH dataset identifies states using the two-digit `FIPSST` state FIPS code.

To support later integration between NSCH and ACS, a matching `FIPSST` field is created from the ACS `GEO_ID`.

This preserves the full ACS geography identifier while also providing a common state-level key across datasets.

```python
# Create state FIPS code from ACS GEO_ID

acs_curated["FIPSST"] = (
    acs_curated["GEO_ID"]
    .str.extract(r"US(\d{2})")[0]
)
```

```python
print("Missing FIPSST values:", acs_curated["FIPSST"].isna().sum())
print("Unique state FIPS codes:", acs_curated["FIPSST"].nunique())

acs_curated[
    ["GEO_ID", "NAME", "FIPSST"]
].head(10)
```
